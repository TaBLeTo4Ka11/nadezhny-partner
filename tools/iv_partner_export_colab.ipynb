# Экспорт контента «Надёжный партнёр»
Этот ноутбук запускает официальный экспортёр WordPress/WooCommerce. Он уважает `robots.txt` и останавливается при блокировках.

**Важно:** не включайте подтверждение разрешения, если владелец сайта явно не разрешил автоматическую выгрузку. Если сайт запрещает автоматический доступ, используйте официальный экспорт WordPress/WooCommerce.


```python
!pip -q install requests beautifulsoup4
import os, getpass, urllib.request, pathlib

# Менять на True можно ТОЛЬКО после явного разрешения владельца сайта.
OWNER_PERMISSION_CONFIRMED = False
os.environ['IVP_OWNER_AUTHORIZED'] = '1' if OWNER_PERMISSION_CONFIRMED else '0'

# Необязательно: отдельные WooCommerce REST API ключи с правами только на чтение.
# Оставьте пустыми, если владелец их не предоставил.
wc_key = getpass.getpass('WooCommerce read-only Consumer Key (Enter — пропустить): ')
wc_secret = getpass.getpass('WooCommerce read-only Consumer Secret (Enter — пропустить): ') if wc_key else ''
if wc_key and wc_secret:
    os.environ['IVP_WC_KEY'] = wc_key
    os.environ['IVP_WC_SECRET'] = wc_secret

wp_user = input('WordPress username (Enter — пропустить): ')
wp_password = getpass.getpass('WordPress Application Password (Enter — пропустить): ') if wp_user else ''
if wp_user and wp_password:
    os.environ['IVP_WP_USER'] = wp_user
    os.environ['IVP_WP_APP_PASSWORD'] = wp_password

script_url = 'https://raw.githubusercontent.com/TaBLeTo4Ka11/nadezhny-partner/gpt6/site-completion/tools/scrape_iv_partner.py'
urllib.request.urlretrieve(script_url, 'scrape_iv_partner.py')
!python scrape_iv_partner.py

archive = pathlib.Path('iv-partner-export.zip')
if archive.exists():
    from google.colab import files
    files.download(str(archive))
else:
    print('Архив не создан. Прочитайте сообщение об остановке и журнал EXPORT_STOPPED.txt, если он был создан.')
```
